# **Project Summary -**

The objective of the project is to design an intelligent chatbot that provides users with information in relation to banking inquiries using large language models (LLMs) and natural language processing (NLP).

An RAG framework will be used along with a pre-trained LLM. The chatbot won't only use the knowledge present in the pre-trained model; however, it will also fetch the information from a corpus of banking documents and FAQs or policies etc. This information would be provided to the LLM as context to generate appropriate responses in response to users' inquiries.

It involves gathering and preparing banking documents, splitting these documents into small text chunks, conversion of these chunks into numerical values through embedding models, and storage of chunks in the vector database. Whenever any inquiry is raised by the user, the relevant text chunks would be fetched and appropriate responses generated using a pre-trained LLM.

The chatbot will be centered on delivering information regarding banking products, policies, processes, and customer service. The development of the system will target an improvement in access to banking information, less effort in accessing relevant information, and making the process more interactive.
The project will involve working on developing a functional chatbot using RAG for Banking queries.

# **Problem Statement**


The banking industry offers various services and customers often ask about banking policies, procedures, fees, and services. Nevertheless, finding correct information about them can be difficult because of the availability of the information in various resources. The problem is that rule-based chatbots cannot understand the customer’s requests and general-purpose Large Language Models might provide incorrect or outdated answers to the banking-domain specific questions. Thus, there is a need for the intelligent banking chatbot capable of understanding customer queries related to banking and giving the correct context-aware answers. This research paper will discuss the development of the banking-specific IndustryGPT which is going to be based on the pre-trained Large Language Model and Retrieval-Augmented Generation (RAG).

```python
from langchain_community.document_loaders import PyPDFLoader

pdf_file = list(uploaded.keys())[0]

loader = PyPDFLoader(pdf_file)

documents = loader.load()

print("Document loaded successfully")
```

```python
# Creating the chunks from uploaded file

from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(chunk_size=500,chunk_overlap=100)

chunks = text_splitter.split_documents(documents)
```

```python
# Embeddings

from sentence_transformers import SentenceTransformer

embedding_model = SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2")

print("Embedding model loaded successfully")

chunk_texts = [chunk.page_content for chunk in chunks]

embeddings = embedding_model.encode(chunk_texts,show_progress_bar=True)

print(f'Embedding shape: {embeddings.shape}')
```

```python
print(embeddings[0])

print("Embedding dimensions:", len(embeddings[0]))
```

```python
# Storing the embeddings

import faiss
import numpy as np

embeddings_np = np.array(embeddings).astype("float32")

dimension = embeddings_np.shape[1]

index = faiss.IndexFlatL2(dimension)

index.add(embeddings_np)

print(dimension)

print("Total vectors stored:", index.ntotal)
```

```python
# Sample user query

query = "What is NEFT?"

query_embedding = embedding_model.encode([query])

query_embedding = np.array(query_embedding).astype("float32")

distances, indices = index.search(query_embedding, k=3)

print("Retrieved chunk indices:", indices)
print("Distances:", distances)

for i in indices[0]:
    print("\nRetrieved Chunk:")
    print(chunks[i].page_content)
```

```python
#LLM

from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

model_name = "google/flan-t5-base"

tokenizer = AutoTokenizer.from_pretrained(model_name)

llm_model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

print("LLM loaded successfully")
```

```python
# Prompt Template

from langchain_core.prompts import PromptTemplate

# Define the RAG prompt template
prompt_template = PromptTemplate(
    input_variables=["context", "question"],
    template="""
You are a helpful banking information assistant.

Answer the user's question using only the banking information
provided in the context below.

If the answer is not available in the context, clearly state
that the provided information is insufficient.

Do not invent banking rules, charges, limits, or policies.

Context:
{context}

Question:
{question}

Answer:
"""
)

print("Prompt template created successfully")
```

```python
# Asking a question

question = input("Ask your banking question: ")

query_embeddings = embedding_model.encode([question])

print(query_embeddings.shape)
```

```python
# Retrieval

distances,indices = index.search(np.array(query_embeddings).astype("float32"),k=3)

retrieved_chunks = [chunks[i].page_content for i in indices[0]]

context = "\n".join(retrieved_chunks)

formatted_prompt = prompt_template.format(context=context,question=question)

print("Prompt created successfully")
```

```python
# LLM generation

inputs = tokenizer(formatted_prompt,return_tensors="pt")

outputs = llm_model.generate(**inputs,max_new_tokens=100)

answer = tokenizer.decode(outputs[0],skip_special_tokens=True)

print("Answer: ",answer)
```

```python
# How to handle information out of PDF

def industry_gpt(question):


    query_embedding = embedding_model.encode([question])

    distances, indices = index.search(
        np.array(query_embedding).astype("float32"),
        k=3
    )

    best_distance = distances[0][0]

    print(f'Best distance: {best_distance}')

    RELEVANCE_THRESHOLD = 1.5

    if best_distance > RELEVANCE_THRESHOLD:
        return "The provided banking information is insufficient to answer this question."

    retrieved_chunks = [
        chunks[i].page_content
        for i in indices[0]
        if i != -1
    ]

    context = "\n\n".join(retrieved_chunks)

    formatted_prompt = prompt_template.format(
        context=context,
        question=question
    )

    inputs = tokenizer(
        formatted_prompt,
        return_tensors="pt"
    )

    outputs = llm_model.generate(
        **inputs,
        max_new_tokens=100,
        do_sample=False
    )

    answer = tokenizer.decode(
        outputs[0],
        skip_special_tokens=True
    )

    return answer
```

```python
question = input("Ask your banking question: ")

answer = industry_gpt(question)

print("Answer:", answer)
```
